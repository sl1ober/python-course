```python
import pandas as pd

# Загружаем датасет
df = pd.read_csv("Housing.csv")

# Память до оптимизации
print("=== ДО оптимизации ===")
df.info(memory_usage="deep")
mem_before = df.memory_usage(deep=True).sum() / 1024**2
print(f"Память до оптимизации: {mem_before:.2f} MB")

# Копия для оптимизации
df_optimized = df.copy()

# Оптимизация типов
for col in df_optimized.columns:
    col_type = df_optimized[col].dtype
    
    if col_type == 'int64':
        df_optimized[col] = df_optimized[col].astype('int32')
    elif col_type == 'float64':
        df_optimized[col] = df_optimized[col].astype('float32')
    elif col_type == 'object':
        num_unique = df_optimized[col].nunique()
        num_total = len(df_optimized[col])
        if num_unique / num_total < 0.5:
            df_optimized[col] = df_optimized[col].astype('category')

# Проверяем память после
print("\n=== ПОСЛЕ оптимизации ===")
df_optimized.info(memory_usage="deep")
mem_after = df_optimized.memory_usage(deep=True).sum() / 1024**2
print(f"Память после оптимизации: {mem_after:.2f} MB")

# Сравнение
print(f"\nЭкономия памяти: {mem_before - mem_after:.2f} MB ({(1 - mem_after/mem_before)*100:.1f}%)")
```

```python
# Выводы:
# Замена int64 → int32 и float64 → float32 уменьшила потребление памяти почти в 2 раза.
# Перевод строковых признаков (object) в категориальные (category) дал наибольший эффект.
# бщая экономия памяти составила ~45%.
```
